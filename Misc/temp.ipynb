```python
import geopandas as gpd
import plotly.offline as pyo
import plotly.graph_objects as go
from plotly.subplots import make_subplots
pyo.init_notebook_mode()


ABT = gpd.read_file("../../../../Data/Final_dataset/ABT/ABT.gpkg", layer="subdivisions")
building_footprint_dataset = gpd.read_file(
    '../../../../Data/Final_dataset/ABT/outputs_building_overlap/Final_Buildings_SemanticIntegrated.gpkg', columns=["geometry"]).to_crs(
    ABT.crs)
meck_bo = gpd.read_file(
    "../../../../Data/Original_dataset/Archive/mecklenburgcounty_boundary/MecklenburgCounty_Boundary.shp").to_crs(
    ABT.crs)

ABT["area_acre"] = ABT.geometry.area / 43560
ABT = ABT[(ABT['year'] >= 1980) & (ABT['year'] <= 2023)]
```

```python
years = list(range(2018, 2024))
abt_filtered = ABT[ABT["year"].isin(years)]

for yr in years:
    data = abt_filtered.loc[abt_filtered["year"] == yr, "BAD"].dropna()

    fig = make_subplots(
        rows=2, cols=1,
        shared_xaxes=True,
        row_heights=[0.7, 0.3],
        vertical_spacing=0.05,
        subplot_titles=[
            f"Histogram of BAD – {yr}",
            "Box Plot"
        ]
    )

    # Histogram
    fig.add_trace(
        go.Histogram(
            x=data,
            nbinsx=30,
            name="Histogram"
        ),
        row=1, col=1
    )

    # Box plot
    fig.add_trace(
        go.Box(
            x=data,
            name="BAD",
            boxpoints="outliers"
        ),
        row=2, col=1
    )

    fig.update_layout(
        title=f"BAD Distribution – {yr}",
        showlegend=False,
        bargap=0.05
    )

    fig.update_xaxes(title_text="BAD", row=2, col=1)
    fig.update_yaxes(title_text="Frequency", row=1, col=1)


    pyo.iplot(fig)
```

```python
import plotly.graph_objects as go

years = list(range(2018, 2024))
abt_filtered = ABT[ABT["year"].isin(years)]

fig = go.Figure()

for yr in years:
    data = abt_filtered.loc[abt_filtered["year"] == yr, "BAD"].dropna()
    fig.add_trace(
        go.Violin(
            y=data,
            name=str(yr),
            box_visible=True,
            meanline_visible=True,
            points="outliers"
        )
    )

fig.update_layout(
    title="BAD Distribution by Year (Violin Plot)",
    yaxis_title="BAD"
)

pyo.iplot(fig)
```

```python
import pandas as pd
import plotly.graph_objects as go

summary = (
    abt_filtered
    .groupby("year")["BAD"]
    .quantile([0.25, 0.5, 0.75])
    .unstack()
    .reset_index()
)

fig = go.Figure()

# IQR band
fig.add_trace(go.Scatter(
    x=summary["year"],
    y=summary[0.75],
    mode="lines",
    line=dict(width=0),
    showlegend=False
))

fig.add_trace(go.Scatter(
    x=summary["year"],
    y=summary[0.25],
    mode="lines",
    fill="tonexty",
    name="IQR",
    line=dict(width=0)
))

# Median
fig.add_trace(go.Scatter(
    x=summary["year"],
    y=summary[0.5],
    mode="lines+markers",
    name="Median BAD"
))

fig.update_layout(
    title="Median BAD with Interquartile Range (2018–2023)",
    yaxis_title="BAD",
    xaxis_title="Year"
)

pyo.iplot(fig)
```

```python
# Imports
# ============================================================
import geopandas as gpd
import folium
from folium.plugins import Search

# Load datasets
# ============================================================

# Building footprints
buildings = building_footprint_dataset

# Pre-processing
# ============================================================

# Guardrail: ensure projected CRS for area calculation
if ABT.crs.is_geographic:
    raise ValueError("ABT must be in a projected CRS to calculate area")

# Area in acres
ABT["area_acre"] = ABT.geometry.area / 43560

# Filter parameters
years = list(range(2018, 2024))
BAD_MIN = 0.10
BAD_MAX = 0.15


# Filter subdivisions by year and BAD
subs_low_bad = (
    ABT[
        (ABT["year"].isin(years)) &
        (ABT["BAD"].between(BAD_MIN, BAD_MAX))
    ]
    .copy()
)

if subs_low_bad.empty:
    raise ValueError("No subdivisions matched the BAD/year criteria")

print(f"Low-BAD subdivisions: {len(subs_low_bad)}")

# Select ONLY buildings within selected subdivisions (FAST)
# ============================================================

buildings_sel = (
    gpd.sjoin(
        buildings,
        subs_low_bad[["geometry"]],
        how="inner",
        predicate="intersects"
    )
    .drop(columns="index_right")
)

print(f"Buildings inside selected subdivisions: {len(buildings_sel)}")

# Prepare data for Folium
# ============================================================

# Reproject to WGS84
subs_low_bad = subs_low_bad.to_crs(epsg=4326)
# buildings_sel = buildings_sel.to_crs(epsg=4326)

# Map center
center_geom = subs_low_bad.unary_union.centroid
center = [center_geom.y, center_geom.x]

# Year colors
year_colors = {
    2018: "#1f77b4",
    2019: "#ff7f0e",
    2020: "#2ca02c",
    2021: "#d62728",
    2022: "#9467bd",
    2023: "#8c564b",
}

# Create Folium map
# ============================================================

m = folium.Map(
    location=center,
    zoom_start=10,
    tiles="CartoDB positron"
)

# ============================================================
# SEARCH LAYER
# ============================================================

search_layer = folium.GeoJson(
    subs_low_bad,
    name="Subdivision Search",
    style_function=lambda x: {
        "fillOpacity": 0,
        "weight": 0,
        "color": "transparent",
    },
    tooltip=folium.GeoJsonTooltip(
        fields=["subd_id"],
        aliases=["Subdivision ID:"]
    )
)

search_layer.add_to(m)

Search(
    layer=search_layer,
    search_label="subd_id",
    placeholder="Search Subdivision ID",
    collapsed=True,
    zoom=14
).add_to(m)

# ============================================================
# Year-based subdivision layers
# ============================================================

def make_style(color):
    return lambda x: {
        "fillColor": color,
        "color": "black",
        "weight": 0.4,
        "fillOpacity": 0.6,
    }

for yr in years:
    subs_yr = subs_low_bad[subs_low_bad["year"] == yr].copy()

    if subs_yr.empty:
        continue

    # Drop only the actual year column
    subs_yr = subs_yr.drop(columns=["year"], errors="ignore")

    fg = folium.FeatureGroup(name=f"Subdivisions {yr}")

    folium.GeoJson(
        subs_yr,
        style_function=make_style(year_colors[yr]),
        tooltip=folium.GeoJsonTooltip(
            fields=["subd_id", "BAD", "area_acre"],
            aliases=["Subdivision ID:", "BAD:", "Area (acres):"],
            localize=True
        )
    ).add_to(fg)

    fg.add_to(m)

# ============================================================
# Building footprints layer
# ============================================================

folium.FeatureGroup(
    name="Building Footprints",
    show=True
).add_child(
    folium.GeoJson(
        buildings_sel,
        style_function=lambda x: {
            "fillColor": "red",
            "color": "red",
            "weight": 0.2,
            "fillOpacity": 0.4,
        }
    )
).add_to(m)

# Layer control
folium.LayerControl(collapsed=False).add_to(m)

m
```

```python
import pandas as pd
from IPython.display import display, HTML

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)

# Clean list of low-BAD subdivisions (table)
# ============================================================

subs_list = (
    subs_low_bad
    .drop(columns="geometry")
    .sort_values(["year", "BAD"])
    .reset_index(drop=True)
)

# Optional: round for cleaner display
subs_list["BAD"] = subs_list["BAD"].round(3)
subs_list["area_acre"] = subs_list["area_acre"].round(2)

# Display full table in scrollable HTML
# ============================================================

display(HTML(
    subs_list
    .to_html(index=False)
    .replace(
        "<table",
        '<table style="display:block; overflow:auto; max-height:600px; width:100%;"'
    )
))
```

```python
import os

output_dir = "../../../../output/low_BAD_ID_lists"
os.makedirs(output_dir, exist_ok=True)

for yr in years:
    df = (
        subs_low_bad
        .loc[subs_low_bad["year"] == yr, ["subd_id", "BAD", "year"]]
        .drop_duplicates()
        .sort_values("subd_id")
        .reset_index(drop=True)
    )

    # Optional: cleaner numbers
    df["BAD"] = df["BAD"].round(4)

    out_path = os.path.join(output_dir, f"low_BAD_subd_ids_{yr}_010_015.csv")
    df.to_csv(out_path, index=False)

    print(f"{yr}: {len(df)} rows written → {out_path}")
```
